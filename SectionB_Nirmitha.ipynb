### Q11
**Prediction:**
```
[[99, 0, 0], [0, 0, 0], [0, 0, 0]]
[[99, 0, 0], [99, 0, 0], [99, 0, 0]]
```
**Why:** `[[0]*3]*3` repeats a reference to the SAME inner list three times, so changing one row changes all. The list comprehension builds three separate lists.

````python
matrix1 = [[0] * 3 for _ in range(3)]
matrix2 = [[0] * 3] * 3
matrix1[0][0] = 99
matrix2[0][0] = 99
print(matrix1)
print(matrix2)
````

### Q12
**Prediction:** `x = [1, 2, 3, 4]`, `y = [1, 2, 3]`, `a = [1, 2, 3, 4]`, `b = [1, 2, 3, 4]`
**Why:** `x + [4]` builds a NEW list and rebinds `x`, so `y` still points to the old one. `a += [4]` extends the list IN PLACE (like `extend`), and `b` points to that same list.

````python
x = [1, 2, 3]
y = x
x = x + [4]
print("x =", x)
print("y =", y)
a = [1, 2, 3]
b = a
a += [4]
print("a =", a)
print("b =", b)
````

### Q13
**Prediction:** first `print(i)` → `4`. `range(0)` is empty, so `j` is never created and `print(j)` raises `NameError: name 'j' is not defined`.
**Why:** a loop variable stays alive after the loop and keeps its last value (4). With zero iterations it is never assigned.

````python
for i in range(5):
    pass
print(i)
for j in range(0):
    pass
try:
    print(j)
except NameError as e:
    print("NameError:", e)
````

### Q14
**Prediction:** `1020`, `30`, `101010`, `3333333333`, then `"10" + 20` raises `TypeError` (cannot join str and int), `30`.
**Why:** `+` joins two strings or adds two numbers but won't mix them; `*` repeats a string; `int("10")` converts first.

````python
print("10" + "20")
print(10 + 20)
print("10" * 3)
print(10 * "3")
try:
    print("10" + 20)
except TypeError as e:
    print("TypeError:", e)
print(int("10") + 20)
````

### Q15
**Prediction:** `5`, `5`, `hello`, `0`, `10`, `default`, `[0]`
**Why (short-circuit):** `or` returns the first truthy operand, `and` returns the first falsy operand (or the last one if all are truthy), and stops evaluating as soon as the answer is known.
`[0]` is a non-empty list, so it is truthy and gets printed, not "default".

````python
print(0 or 5)
print(5 or 0)
print(None or "hello")
print(0 and 5)
print(5 and 10)
print([] or "default")
print([0] or "default")
````
