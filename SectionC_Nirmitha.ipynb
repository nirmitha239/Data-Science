### Q16. Swap two numbers without a third variable

```python
a, b = 5, 9
print("Before:", a, b)

# (a) Arithmetic
a = a + b      # a holds the sum
b = a - b      # sum - b = original a
a = a - b      # sum - original a = original b
print("Arithmetic:", a, b)

# (b) XOR (works only for integers)
a, b = 5, 9
a = a ^ b
b = a ^ b
a = a ^ b
print("XOR       :", a, b)

# (c) Tuple unpacking
a, b = 5, 9
a, b = b, a
print("Unpacking :", a, b)

# Most Pythonic: (c). It is readable, works for any type (str, float, list...),
# and has no overflow/type restrictions. (a) fails to be exact for floats,
# (b) works only for ints.
```

### Q21. Income Tax Calculator (New Regime, FY 2024-25, slab-wise)

```python
income = float(input("Annual income (₹): "))

# (lower limit, upper limit, rate)
slabs = [
    (300000, 700000, 0.05),
    (700000, 1000000, 0.10),
    (1000000, 1200000, 0.15),
    (1200000, 1500000, 0.20),
    (1500000, float("inf"), 0.30),
]

total_tax = 0
print(f"Up to ₹3,00,000: Nil")
for lower, upper, rate in slabs:
    if income > lower:
        taxable = min(income, upper) - lower   # part of income falling in this slab
        tax = taxable * rate
        total_tax += tax
        print(f"₹{taxable:,.0f} @ {rate * 100:.0f}% = ₹{tax:,.0f}")
print(f"Total tax: ₹{total_tax:,.0f}")
```

### Q22. PAN validator (no regex)

```python
pan = input("Enter PAN: ").strip()

def is_upper_letter(ch):
    return "A" <= ch <= "Z"

def is_digit(ch):
    return "0" <= ch <= "9"

reason = None
if len(pan) != 10:
    reason = f"length must be 10, got {len(pan)}"
else:
    for i in range(5):
        if not is_upper_letter(pan[i]):
            reason = f"character {i + 1} ('{pan[i]}') must be an uppercase letter"
            break
    if reason is None:
        for i in range(5, 9):
            if not is_digit(pan[i]):
                reason = f"character {i + 1} ('{pan[i]}') must be a digit"
                break
    if reason is None and not is_upper_letter(pan[9]):
        reason = f"character 10 ('{pan[9]}') must be an uppercase letter"

print("Valid PAN" if reason is None else f"Invalid PAN — {reason}")
```

### Q27. Number diamond

```python
N = int(input("Enter N: "))

def print_row(i, n):
    print("  " * (n - i), end="")          # leading spaces (2 per level)
    for j in range(1, i + 1):              # ascending
        print(j, end=" ")
    for j in range(i - 1, 0, -1):          # descending
        print(j, end=" ")
    print()

for i in range(1, N + 1):          # upper half including middle
    print_row(i, N)
for i in range(N - 1, 0, -1):      # lower half
    print_row(i, N)
```

### Q29. Pascal's Triangle (first 5 rows)

```python
rows = 5
for r in range(rows):
    print("  " * (rows - 1 - r), end="")
    value = 1
    for k in range(r + 1):
        print(value, end=" ")
        value = value * (r - k) // (k + 1)   # C(r, k+1) from C(r, k)
    print()
```

### Q31. Armstrong numbers 1 to 9999

```python
armstrong = []
for num in range(1, 10000):
    # count digits
    digits = 0
    temp = num
    while temp > 0:
        digits += 1
        temp //= 10
    # sum of digits ** digits
    total = 0
    temp = num
    while temp > 0:
        total += (temp % 10) ** digits
        temp //= 10
    if total == num:
        armstrong.append(num)

print(", ".join(str(x) for x in armstrong))
print("Count:", len(armstrong))
```

### Q33. Optimized prime check and sum of primes up to N

```python
def is_prime(num):
    if num < 2:
        return False
    i = 2
    while i * i <= num:       # only up to sqrt(num)
        if num % i == 0:
            return False
        i += 1
    return True

N = int(input("Enter N: "))
print(f"{N} is {'prime' if is_prime(N) else 'not prime'}")

prime_sum = 0
for k in range(2, N + 1):
    if is_prime(k):
        prime_sum += k
print(f"Sum of primes from 2 to {N}: {prime_sum}")
```

### Q34. Reverse a number and check palindrome (no strings)

```python
original = int(input("Enter a number: "))
if original < 0:
    print(f"{original} → NOT a palindrome (negative sign)")
else:
    temp = original
    reversed_num = 0
    while temp > 0:
        reversed_num = reversed_num * 10 + temp % 10
        temp //= 10
    print("Reversed:", reversed_num)
    print("Palindrome" if reversed_num == original else "Not a palindrome")
```

### Q35. HCF (Euclid) and LCM

```python
x = int(input("First positive integer: "))
y = int(input("Second positive integer: "))

a, b = x, y
while b != 0:
    a, b = b, a % b       # gcd(a, b) = gcd(b, a mod b)
hcf = a
lcm = x * y // hcf        # HCF * LCM = a * b
print("HCF:", hcf)
print("LCM:", lcm)
```

### Q36. Single-pass string scanner

```python
text = input("Enter a sentence: ")

total_chars = words = vowels = consonants = digits = spaces = special = 0
in_word = False

for ch in text:
    total_chars += 1
    if ch == " ":
        spaces += 1
        in_word = False
    else:
        if not in_word:          # a new word starts here
            words += 1
            in_word = True
        low = ch.lower()
        if "a" <= low <= "z":
            if low in "aeiou":
                vowels += 1
            else:
                consonants += 1
        elif "0" <= ch <= "9":
            digits += 1
        else:
            special += 1

print("Total characters:", total_chars)
print("Words:", words)
print("Vowels:", vowels)
print("Consonants:", consonants)
print("Digits:", digits)
print("Spaces:", spaces)
print("Special characters:", special)
```

### Q37. Student marks analysis (no sort/sorted)

```python
marks = list(map(int, input("Enter 10 marks (space-separated): ").split()))

maximum = minimum = marks[0]
total = 0
first = second = None          # highest and second-highest distinct values
for m in marks:
    total += m
    if m > maximum:
        maximum = m
    if m < minimum:
        minimum = m
    if first is None or m > first:
        second = first
        first = m
    elif m != first and (second is None or m > second):
        second = m

average = total / len(marks)
above_avg = 0
for m in marks:
    if m > average:
        above_avg += 1

print("Maximum:", maximum)
print("Minimum:", minimum)
print("Average:", average)
print("Above average:", above_avg)
print("Second highest:", second if second is not None else "N/A (all marks equal)")
```

### Q38. Word Frequency Analyzer

```python
review = input("Enter a review: ")

# clean punctuation, lower-case
cleaned = ""
for ch in review.lower():
    if ("a" <= ch <= "z") or ("0" <= ch <= "9") or ch == " ":
        cleaned += ch

freq = {}
for word in cleaned.split():
    freq[word] = freq.get(word, 0) + 1

# sort keys by their count, descending (ties keep first-seen order)
for word in sorted(freq, key=freq.get, reverse=True):
    print(f"{word}: {freq[word]}")

top_word = max(freq, key=freq.get)
print(f"Most frequent: '{top_word}' ({freq[top_word]} times)")
```

### Q41. TCS weekly stock analysis

```python
prices = (3845, 3902, 3878, 3955, 4012, 3998, 4055)
days = ("Mon", "Tue", "Wed", "Thu", "Fri", "Sat", "Sun")

max_i = min_i = 0
total = 0
best_gain = prices[1] - prices[0]
best_gain_i = 1
volatility = 0
for i in range(len(prices)):
    total += prices[i]
    if prices[i] > prices[max_i]:
        max_i = i
    if prices[i] < prices[min_i]:
        min_i = i
    if i > 0:
        change = prices[i] - prices[i - 1]
        volatility += change if change >= 0 else -change
        if change > best_gain:
            best_gain, best_gain_i = change, i

print(f"(a) Max: ₹{prices[max_i]} on {days[max_i]}")
print(f"(b) Min: ₹{prices[min_i]} on {days[min_i]}")
print(f"(c) Average: ₹{total / len(prices):.2f}")
print(f"(d) Biggest gain: ₹{best_gain} ({days[best_gain_i - 1]} → {days[best_gain_i]})")
print(f"(e) Total volatility: ₹{volatility}")
```

### Q43. Smart Palindrome Checker

```python
sentence = input("Enter a sentence: ")

cleaned = ""
for ch in sentence.lower():
    if ("a" <= ch <= "z") or ("0" <= ch <= "9"):
        cleaned += ch

# two-pointer check
left, right = 0, len(cleaned) - 1
is_pal = True
while left < right:
    if cleaned[left] != cleaned[right]:
        is_pal = False
        break
    left += 1
    right -= 1
print("Palindrome" if is_pal else "Not a palindrome")
```

### Q45. Caesar Cipher

```python
message = input("Message: ")
shift = int(input("Shift: ")) % 26     # wraps big or negative shifts

encrypted = ""
for ch in message:
    if "A" <= ch <= "Z":
        encrypted += chr((ord(ch) - ord("A") + shift) % 26 + ord("A"))
    elif "a" <= ch <= "z":
        encrypted += chr((ord(ch) - ord("a") + shift) % 26 + ord("a"))
    else:
        encrypted += ch                # digits, spaces, symbols unchanged
print(encrypted)
```

### Q46. calculate_gst

```python
def calculate_gst(amount, gst_rate=18, gst_inclusive=False):
    """Return (base_amount, gst_amount, total_amount), each rounded to 2 decimals."""
    if gst_inclusive:
        base = amount / (1 + gst_rate / 100)   # strip GST out of the total
        gst = amount - base
        total = amount
    else:
        base = amount
        gst = amount * gst_rate / 100
        total = amount + gst
    return (round(base, 2), round(gst, 2), round(total, 2))

print(calculate_gst(1000))
print(calculate_gst(1180, gst_inclusive=True))
print(calculate_gst(500, gst_rate=5))
```

### Q47. Recursive power

```python
def power(base, exp):
    if exp == 0:
        return 1
    if exp < 0:
        return 1 / power(base, -exp)
    return base * power(base, exp - 1)

print(power(2, 10))
print(power(5, 0))
print(power(3, -2))
```
