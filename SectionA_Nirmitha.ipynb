# Section A: Theory (Q1 to Q10)

## Section A: Theory

**Q1. `is` vs `==`**
`==` checks whether two objects have the same *value*. `is` checks whether they are the *same object in memory*.
Example: `a = [1, 2, 3]`, `b = [1, 2, 3]` → `a == b` is True, `a is b` is False (two separate lists).

**Q2. Mutable vs Immutable**
Mutable: list, set, dict. Immutable: tuple, str, int, float, bool, frozenset.
Immutability matters because the value can never change unexpectedly, so it is safe to share and can be used as a dict key or set member.

**Q3. `print(0.1 + 0.2 == 0.3)`**
Output: `False`. Computers store floats in binary, and 0.1 and 0.2 cannot be stored exactly, so the sum is 0.30000000000000004.
It is not a Python bug; every language using IEEE-754 floats does this. Compare with `abs(a - b) < 1e-9` instead.

**Q4. Tuple over list (3 situations)**
1. Fixed data such as coordinates `(12.97, 77.59)` or RGB colours: it should never change, and a tuple enforces that.
2. Dictionary keys or set members: tuples are hashable, lists are not.
3. Returning multiple values from a function, or protecting data from accidental edits; tuples are also a little lighter and faster.

**Q5. Dict lookup O(1) vs list `in` O(n)**
*Without "hash":* a dict turns the key into a number that tells it exactly which slot to look in, so it jumps straight there. A list has no such shortcut, so it checks items one by one from the start.
*With "hash":* a dict applies `hash(key)` to get an integer, maps it to a bucket index and goes directly there (average constant time). A list `in` compares against every element, so time grows with length.

**Q6. `range()` vs `numpy.arange()`**
1. `range` accepts only integers; `arange` accepts floats, e.g. `np.arange(0, 1, 0.1)` works, `range(0, 1, 0.1)` raises TypeError.
2. `range` returns a lazy range object; `arange` returns an actual ndarray stored in memory.
3. `arange` output supports vectorized maths (`arr * 2`) and has a dtype; `range` does not.

**Q7. append vs extend vs insert** (with `a = [1, 2, 3]`, `x = [4, 5]`)
- `a.append(x)` → `[1, 2, 3, [4, 5]]` (adds x as ONE element)
- `a.extend(x)` → `[1, 2, 3, 4, 5]` (adds each item of x)
- `a.insert(1, x)` → `[1, [4, 5], 2, 3]` (puts x as one element at index 1)

**Q8. Extended unpacking**
`a, *b, c = [10, 20, 30, 40, 50]` → `a = 10`, `b = [20, 30, 40]`, `c = 50`. First and last go to a and c, the middle goes into a list.
`a, *b, c = [10, 20]` → `a = 10`, `b = []`, `c = 20`.
`a, *b, c = [10]` → `ValueError: not enough values to unpack` (needs at least 2 items for a and c).

**Q9. `/`, `//`, `%`, `**` on (-7, 2)**
- `-7 / 2` = `-3.5` (true division)
- `-7 // 2` = `-4` (floor division rounds DOWN, not toward zero)
- `-7 % 2` = `1` (remainder has the sign of the divisor; check: -4 × 2 + 1 = -7)
- `(-7) ** 2` = `49` (power). Note `-7 ** 2` without brackets is `-49`, because `**` binds tighter than the minus.

**Q10. Why NumPy is faster than lists**
1. Memory layout: an array stores same-type values in one contiguous block; a list stores pointers to separate objects scattered in memory, which is slower and not cache-friendly.
2. Vectorized operations run in compiled C loops, avoiding the Python interpreter overhead of looping element by element.
3. Fixed dtype means no per-element type checking, and uses less memory (e.g. int32 vs full Python int objects).
