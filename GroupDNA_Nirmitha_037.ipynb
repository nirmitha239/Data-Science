## Feature 1: The Chat Parser
Reads the export line by line. Handles system messages, `<Media omitted>`, deleted messages, multi-line messages and empty lines.

```python
# AI-assisted: structure of this cell was drafted with Claude's help; I reviewed and understood each line.
from datetime import datetime, timedelta
import numpy as np

# Try the usual places where the dataset may be (local folder or Colab's /content)
DATA_CANDIDATES = ['hostel_bois.txt', '/content/hostel_bois.txt', 'hostel_bois_practice.txt', '/content/hostel_bois_practice.txt']

def load_lines():
    for path in DATA_CANDIDATES:
        try:
            with open(path, 'r', encoding='utf-8') as f:
                print(f'Reading file: {path}')
                return f.read().split('\n')
        except FileNotFoundError:
            continue
    print('Chat file not found. Upload hostel_bois.txt to this Colab session.')
    return []

def has_date_prefix(line):
    """True if the line starts with NN/NN/NN (first 8 characters), without regex."""
    if len(line) < 8:
        return False
    for i in (0, 1, 3, 4, 6, 7):
        if not line[i].isdigit():
            return False
    return line[2] == '/' and line[5] == '/'

def parse_chat(lines):
    """Return (messages, info). Media and deleted entries are kept (they were sent!) but tagged with a kind."""
    messages = []                       # every non-system message (text, media, deleted), each with a 'kind'
    info = {'system': 0, 'media': {}, 'deleted': {}, 'group': 'Unknown group'}
    current = None                      # last parsed message (for multi-line continuation)

    for raw in lines:
        line = raw.rstrip('\r')
        if line.strip() == '':          # empty lines: skip silently
            continue

        if not has_date_prefix(line):   # continuation of previous message
            if current is not None:
                current['text'] += ' ' + line.strip()
            continue

        if ' - ' not in line:           # cannot split into (timestamp, rest) -> system line
            info['system'] += 1
            current = None
            continue
        timestamp, rest = line.split(' - ', 1)      # split only once
        if ': ' not in rest:            # no "sender: text" -> system message
            info['system'] += 1
            current = None
            if ' created group "' in rest:
                info['group'] = rest.split('"')[1]
            continue

        sender, text = rest.split(': ', 1)
        text = text.strip()
        kind = 'text'
        if text == '<Media omitted>':
            kind = 'media'
            info['media'][sender] = info['media'].get(sender, 0) + 1
        elif text == 'This message was deleted':
            kind = 'deleted'
            info['deleted'][sender] = info['deleted'].get(sender, 0) + 1
        new_msg = {'timestamp': timestamp, 'sender': sender, 'text': text, 'kind': kind,
                   'dt': datetime.strptime(timestamp, '%d/%m/%y, %H:%M')}
        messages.append(new_msg)
        current = new_msg if kind == 'text' else None     # only text messages can continue on the next line
    return messages, info

lines = load_lines()
messages, info = parse_chat(lines)

people = set()
for m in messages:
    people.add(m['sender'])
if messages:
    num_days = (messages[-1]['dt'].date() - messages[0]['dt'].date()).days + 1
else:
    num_days = 0
total_media = sum(info['media'].values())
total_deleted = sum(info['deleted'].values())

print(f"Successfully parsed {len(messages)} messages from {len(people)} participants over {num_days} days, "
      f"skipped {info['system']} system messages, {total_media} media-omitted, {total_deleted} deleted messages.")
print('(Media and deleted entries are counted as messages from the sender, but left out of word and length statistics.)')
print('\nFirst 5 parsed messages:')
for m in messages[:5]:
    print('  ', m['timestamp'], '|', m['sender'], '|', m['text'][:50])
print('Last 5 parsed messages:')
for m in messages[-5:]:
    print('  ', m['timestamp'], '|', m['sender'], '|', m['text'][:50])
```

## Feature 2: Group Overview

```python
# AI-assisted: Claude helped me with the f-string alignment and the bar helper; logic reviewed by me.
WIDTH = 60

def make_bar(value, max_value, width=20):
    """Bar of block characters proportional to value. Tiny values show a dot."""
    if max_value == 0:
        return '.'
    length = round(value / max_value * width)
    return '█' * length if length > 0 else '.'

def text_only(msgs):
    """Only real typed messages (no media stubs, no deleted messages)."""
    out = []
    for m in msgs:
        if m['kind'] == 'text':
            out.append(m)
    return out

def count_per_person(msgs):
    counts = {}
    for m in msgs:
        counts[m['sender']] = counts.get(m['sender'], 0) + 1
    return counts

def ranked(counts):
    """People sorted by message count, highest first."""
    return sorted(counts, key=counts.get, reverse=True)

def print_overview(msgs, info):
    counts = count_per_person(msgs)
    first, last = msgs[0]['dt'], msgs[-1]['dt']
    days = (last.date() - first.date()).days + 1
    total = len(msgs)
    print('=' * WIDTH)
    print(' GROUP OVERVIEW')
    print('=' * WIDTH)
    print(f" Group          : {info['group']}")
    print(f" Period         : {first.strftime('%d %B %Y')} to {last.strftime('%d %B %Y')} ({days} days)")
    print(f' Total messages : {total:,}')
    print(f' Participants   : {len(counts)}')
    print('\n MESSAGES PER PERSON')
    top = max(counts.values())
    for name in ranked(counts):
        c = counts[name]
        print(f' {name:<8} {make_bar(c, top):<20} {c:>5} ({c / total * 100:4.1f}%)')

# Extra per-person stats (Day 2): word count and average message length
def person_text_stats(msgs):
    stats = {}
    for m in msgs:
        words = len(m['text'].split())
        s = stats.setdefault(m['sender'], {'msgs': 0, 'words': 0})
        s['msgs'] += 1
        s['words'] += words
    return stats

print_overview(messages, info)
print('\n AVERAGE MESSAGE LENGTH')
stats = person_text_stats(text_only(messages))
for name in ranked(count_per_person(messages)):
    s = stats[name]
    print(f" {name:<8} {s['words'] / s['msgs']:5.1f} words per message ({s['words']:,} total words)")
```

## Feature 4: Activity Heatmap (NumPy)
A 6 × 24 matrix: rows = people, columns = hour of day. Each cell = total messages sent by that person in that hour.

```python
# AI-assisted: Claude helped with the shading-level logic; the NumPy matrix part I wrote and tested myself.
def build_heatmap(msgs):
    counts = count_per_person(msgs)
    names = ranked(counts)
    row_of = {}
    for i, name in enumerate(names):
        row_of[name] = i
    grid = np.zeros((len(names), 24), dtype=int)     # the NumPy matrix
    for m in msgs:
        grid[row_of[m['sender']], m['dt'].hour] += 1
    return names, grid

def shade(value, row_max):
    """Four shading levels relative to the person's own maximum."""
    if row_max == 0:
        return '. '
    ratio = value / row_max
    if ratio < 0.25:
        return '. '
    if ratio < 0.50:
        return '░ '
    if ratio < 0.75:
        return '▒ '
    return '█ '

def print_heatmap(names, grid, tags=None):
    tags = tags or {}
    print(' ACTIVITY HEATMAP (hour of day, columns 00 to 23)')
    header = ''
    for h in range(24):
        header += f'{h:02d}' if h % 3 == 0 else '  '
    print(f" {'':<8} {header}")
    for i, name in enumerate(names):
        row_max = grid[i].max()
        cells = ''
        for h in range(24):
            cells += shade(grid[i, h], row_max)
        print(f" {name:<8} {cells}{tags.get(name, '')}")

names_by_activity, heatmap = build_heatmap(messages)
print_heatmap(names_by_activity, heatmap)
print('\nShape of matrix:', heatmap.shape)
print('Row totals match message counts:', list(heatmap.sum(axis=1)) == [count_per_person(messages)[n] for n in names_by_activity])
print('Night hours (23,0,1,2,3,4) share per person:')
night_cols = [23, 0, 1, 2, 3, 4]
for i, name in enumerate(names_by_activity):
    print(f'  {name:<8} {heatmap[i, night_cols].sum() / heatmap[i].sum() * 100:5.1f}%')
```

## Feature 5: Top Words

```python
# AI-assisted: Claude suggested the stop-word list idea; the tokenizing and counting loop is mine.
STOP_WORDS = {'i', 'is', 'the', 'a', 'an', 'and', 'or', 'to', 'of', 'in', 'on', 'for', 'it', 'that', 'we',
              'with', 'so', 'but', 'then', 'when', 'at', 'was', 'are', 'am', 'be', 'me', 'my', 'you', 'this',
              'hai', 'hain', 'hoon', 'ho', 'ka', 'ki', 'ke', 'se', 'raha', 'tha', 'how', 'about', 'today', 'he', 'his',
              'she', 'her', 'they', 'them', 'their', 'have', 'has', 'had', 'just', 'which', 'from', 'up', 'one', 'no',
              'not', 'do', 'did', 'can', 'will', 'what', 'who', 'there', 'been', 'were', 'if', 'as', 'by', 'out', 'all',
              'also', 'into', 'after', 'before', 'than', 'too', 'very', 'your', 'its', 'now', 'why', 'off', 'used',
              'came', 'three', 'everyone', 'anyone', 'everything', 'telling', 'started', 'entire'}
# common English / Hindi filler words are removed so the group's own slang (bhai, scene, yaar...) shows up
PUNCTUATION = '.,!?;:"()[]{}-_*/\\<>@#~|&^%$+=`\''

def tokenize(text):
    """Lower-case, strip punctuation from both ends, keep only alphabetic words."""
    words = []
    for raw in text.lower().split():
        word = raw.strip(PUNCTUATION)
        if word.replace("'", '').isalpha():
            words.append(word)
    return words

def word_frequencies(msgs):
    freq = {}
    for m in text_only(msgs):
        for word in tokenize(m['text']):
            if word not in STOP_WORDS:
                freq[word] = freq.get(word, 0) + 1
    return freq

def top_n(freq, n):
    return sorted(freq, key=freq.get, reverse=True)[:n]

def print_top_words(msgs, n=10):
    freq = word_frequencies(msgs)
    top = top_n(freq, n)
    print(" THIS GROUP'S FAVOURITE WORDS")
    for word in top:
        print(f' {word:<10} {make_bar(freq[word], freq[top[0]]):<20} {freq[word]}')

print_top_words(messages)
print('\n TOP 5 WORDS PER PERSON')
for name in ranked(count_per_person(messages)):
    own = []
    for m in messages:
        if m['sender'] == name:
            own.append(m)
    f = word_frequencies(own)
    print(f" {name:<8} " + ', '.join(f'{w} ({f[w]})' for w in top_n(f, 5)))
```

## Feature 6: Response Speed and Silent Streaks

```python
# AI-assisted: Claude helped me with the strptime/timedelta idea; the streak loop I wrote and traced by hand.
def response_times(msgs):
    """Average seconds between someone else's last message and this person's next message."""
    gaps = {}
    for i in range(1, len(msgs)):
        if msgs[i]['sender'] != msgs[i - 1]['sender']:
            seconds = (msgs[i]['dt'] - msgs[i - 1]['dt']).total_seconds()
            gaps.setdefault(msgs[i]['sender'], []).append(seconds)
    averages = {}
    for name in gaps:
        averages[name] = sum(gaps[name]) / len(gaps[name])
    return averages

def format_gap(seconds):
    return f'{seconds / 60:.1f} minutes' if seconds < 3600 else f'{seconds / 3600:.1f} hours'

def silent_streaks(msgs):
    """For each person: (longest streak in days, start date, end date) of consecutive days with 0 messages."""
    first_day, last_day = msgs[0]['dt'].date(), msgs[-1]['dt'].date()
    total_days = (last_day - first_day).days + 1
    active = {}
    for m in msgs:
        active.setdefault(m['sender'], set()).add(m['dt'].date())
    result = {}
    for name in active:
        best, best_start, current, current_start = 0, None, 0, None
        for offset in range(total_days):
            day = first_day + timedelta(days=offset)
            if day in active[name]:
                current = 0
            else:
                if current == 0:
                    current_start = day
                current += 1
                if current > best:
                    best, best_start = current, current_start
        best_end = best_start + timedelta(days=best - 1) if best > 0 else None
        result[name] = (best, best_start, best_end, total_days - len(active[name]), total_days)
    return result

def print_response_patterns(msgs):
    avg = response_times(msgs)
    print(' RESPONSE PATTERNS')
    if len(avg) == 0:      # edge case: only one person, nobody to reply to
        print(' Not enough back-and-forth to measure response times.')
    else:
        fastest = min(avg, key=avg.get)
        slowest = max(avg, key=avg.get)
        print(f' Fastest replier : {fastest} (avg {format_gap(avg[fastest])})')
        print(f' Slowest replier : {slowest} (avg {format_gap(avg[slowest])})')
    streaks = silent_streaks(msgs)
    print('\n LONGEST SILENT STREAKS (consecutive days with zero messages)')
    for name in sorted(streaks, key=lambda n: streaks[n][0], reverse=True):
        days, start, end = streaks[name][0], streaks[name][1], streaks[name][2]
        if days == 0:
            print(f' {name:<8} : 0 days (never went silent)')
        else:
            when = start.strftime('%d %b') if days == 1 else f"{start.strftime('%d %b')} - {end.strftime('%d %b')}"
            print(f" {name:<8} : {days} {'day' if days == 1 else 'days'} ({when})")

print_response_patterns(messages)
```

## Feature 7: Personality Archetype Detection
One scoring function per archetype. Every score is **normalised so that 1.0 = the rule's threshold is just met**, which makes scores comparable across archetypes.

**Tie-breaking rule (exclusive assignment):** list every (score, person, archetype) combination, sort from highest score to lowest, and give an archetype to a person only if neither the person nor the archetype is already taken. So if two people pass the same threshold, the one with the higher score keeps it and the other falls back to their next best archetype.

**Two rounds:** first the six main archetypes (Spammer, Group Mom, Night Owl, Storyteller, Drama Queen, Ghost) are handed out. Only people still without an archetype get one of the fallback ones (Comedian, Question Master, Hype Beast), exactly as the brief calls them tiebreakers.

Archetypes based on "highest count" (Group Mom, Comedian, Hype Beast) have no fixed threshold, so they are scored relative to the group's best (the best person scores 1.0).

```python
# AI-assisted: Claude helped me design the normalised-score idea and the greedy assignment; thresholds come from the brief.
CARING_KEYWORDS = ['okay', 'safe', 'eat', 'sleep', 'take care', 'are you', 'please', 'reminder', 'drink water', "don't forget"]
LAUGH_WORDS = {'lol', 'lmao', 'haha', 'rofl', 'lmfao'}
HYPE_WORDS = {'fire', 'lit', 'op', 'legend', 'bro', 'dhamaal', 'mast'}
NIGHT_HOURS = {23, 0, 1, 2, 3, 4}

def count_keywords(msgs, keywords):
    """Count whole-word / whole-phrase matches (so 'eat' does not match 'great')."""
    total = 0
    for m in msgs:
        low = m['text'].lower()
        for ch in PUNCTUATION.replace("'", ''):     # keep apostrophes: "don't forget"
            low = low.replace(ch, ' ')
        padded = ' ' + ' '.join(low.split()) + ' '
        for key in keywords:
            total += padded.count(' ' + key + ' ')
    return total

def bursts_per_person(msgs):
    """Lengths of runs of back-to-back messages by the same sender."""
    runs = {}
    run_len = 1
    for i in range(1, len(msgs) + 1):
        if i < len(msgs) and msgs[i]['sender'] == msgs[i - 1]['sender']:
            run_len += 1
        else:
            runs.setdefault(msgs[i - 1]['sender'], []).append(run_len)
            run_len = 1
    return runs

# --- one function per archetype: returns (score, human-readable detail)
def score_spammer(own, ctx):
    avg = sum(ctx['bursts']) / len(ctx['bursts'])
    return avg / 3, f'avg {avg:.1f} msgs in a row'

def score_group_mom(own, ctx):
    count = count_keywords(own, CARING_KEYWORDS)
    ctx['caring'] = count
    return (count / ctx['max_caring'] if ctx['max_caring'] else 0), f'caring keyword score: {count}'

def score_night_owl(own, ctx):
    night = 0
    for m in own:
        if m['dt'].hour in NIGHT_HOURS:
            night += 1
    pct = night / len(own) * 100
    return pct / 60, f'{pct:.1f}% msgs between 23h-04h'

def score_storyteller(own, ctx):
    own = text_only(own)
    if len(own) == 0:
        return 0, 'no typed messages'
    words = 0
    for m in own:
        words += len(m['text'].split())
    avg = words / len(own)
    return avg / 30, f'avg {avg:.1f} words per msg'

def score_drama_queen(own, ctx):
    own = text_only(own)
    if len(own) == 0:
        return 0, 'no typed messages'
    loud, caps = 0, 0
    for m in own:
        is_caps = len(m['text']) >= 3 and m['text'].isupper()
        if is_caps:
            caps += 1
        if is_caps or m['text'].count('!') >= 2:
            loud += 1
    pct = loud / len(own) * 100
    caps_pct = caps / len(own) * 100
    detail = f'{caps_pct:.1f}% ALL-CAPS messages' if caps_pct >= 30 else f'{pct:.1f}% loud messages'
    return pct / 30, detail

def score_ghost(own, ctx):
    silent, total = ctx['silent_days'], ctx['total_days']
    return (silent / total * 100) / 60, f'silent on {silent} of {total} days'

def score_comedian(own, ctx):
    own = text_only(own)
    if len(own) == 0:
        return 0, 'no typed messages'
    laughs = 0
    for m in own:
        for word in tokenize(m['text']):
            if word in LAUGH_WORDS or word.startswith('haha'):
                laughs += 1
    pct = laughs / len(own) * 100
    ctx['laugh_pct'] = pct
    return (pct / ctx['max_laugh'] if ctx['max_laugh'] else 0), f'{pct:.1f}% msgs have laughter'

def score_question_master(own, ctx):
    own = text_only(own)
    if len(own) == 0:
        return 0, 'no typed messages'
    questions = 0
    for m in own:
        if m['text'].endswith('?'):
            questions += 1
    pct = questions / len(own) * 100
    return pct / 25, f'{pct:.1f}% msgs end with ?'

# BONUS 9th archetype: THE HYPE BEAST (Indian college slang for excitement)
# Rule: highest share of messages using hype words like 'fire', 'lit', 'legend', 'dhamaal', 'mast'.
# It is a fallback archetype, so it is only given to someone who did not get one of the six main ones.
def score_hype_beast(own, ctx):
    own = text_only(own)
    if len(own) == 0:
        return 0, 'no typed messages'
    hype = 0
    for m in own:
        for word in tokenize(m['text']):
            if word in HYPE_WORDS:
                hype += 1
                break
    pct = hype / len(own) * 100
    ctx['hype_pct'] = pct
    return (pct / ctx['max_hype'] if ctx['max_hype'] else 0), f'{pct:.1f}% msgs use hype words'

# Tier 1 = the six main archetypes; Tier 2 = fallbacks (the brief calls Comedian and Question Master tiebreakers)
MAIN_ARCHETYPES = [
    ('THE SPAMMER', score_spammer), ('THE GROUP MOM', score_group_mom), ('THE NIGHT OWL', score_night_owl),
    ('THE STORYTELLER', score_storyteller), ('THE DRAMA QUEEN', score_drama_queen), ('THE GHOST', score_ghost),
]
FALLBACK_ARCHETYPES = [
    ('THE COMEDIAN', score_comedian), ('THE QUESTION MASTER', score_question_master), ('THE HYPE BEAST', score_hype_beast),
]
ARCHETYPES = MAIN_ARCHETYPES + FALLBACK_ARCHETYPES
FALLBACK_NAMES = [label for label, func in FALLBACK_ARCHETYPES]

def detect_archetypes(msgs):
    people_list = ranked(count_per_person(msgs))
    own_msgs = {name: [] for name in people_list}
    for m in msgs:
        own_msgs[m['sender']].append(m)
    bursts = bursts_per_person(msgs)
    streaks = silent_streaks(msgs)

    # group-wide maximums needed by the "highest count" archetypes
    maxima = {'max_caring': 0, 'max_laugh': 0, 'max_hype': 0}
    for name in people_list:
        maxima['max_caring'] = max(maxima['max_caring'], count_keywords(own_msgs[name], CARING_KEYWORDS))
        l, h = 0, 0
        typed = text_only(own_msgs[name])
        for m in typed:
            toks = tokenize(m['text'])
            if any(w in LAUGH_WORDS or w.startswith('haha') for w in toks):
                l += 1
            if any(w in HYPE_WORDS for w in toks):
                h += 1
        if len(typed) > 0:
            maxima['max_laugh'] = max(maxima['max_laugh'], l / len(typed) * 100)
            maxima['max_hype'] = max(maxima['max_hype'], h / len(typed) * 100)

    table = []                 # (score, person, archetype, detail)
    for name in people_list:
        ctx = dict(maxima)
        ctx['bursts'] = bursts[name]
        ctx['silent_days'], ctx['total_days'] = streaks[name][3], streaks[name][4]
        for label, func in ARCHETYPES:
            score, detail = func(own_msgs[name], ctx)
            table.append((score, name, label, detail))

    # exclusive assignment: highest scores first (see tie-breaking note above)
    table_sorted = sorted(table, key=lambda row: -row[0])
    assigned, taken = {}, set()
    for tier_two in (False, True):          # round 1: main six, round 2: fallbacks for anyone left
        for score, name, label, detail in table_sorted:
            is_fallback = label in FALLBACK_NAMES
            if is_fallback != tier_two:
                continue
            if score > 0 and name not in assigned and label not in taken:
                assigned[name] = (label, detail, score)
                taken.add(label)
    for name in people_list:
        if name not in assigned:
            assigned[name] = ('THE MYSTERY MEMBER', 'not enough signal', 0)

    # bonus: runner-up archetype (best score among archetypes that are not their own)
    runner_up = {}
    for score, name, label, detail in table_sorted:
        if name not in runner_up and label != assigned[name][0] and score > 0:
            runner_up[name] = label
    return people_list, assigned, runner_up

def print_archetypes(msgs, show_runner_up=True):
    people_list, assigned, runner_up = detect_archetypes(msgs)
    print(' PERSONALITY ARCHETYPES')
    for name in people_list:
        label, detail, score = assigned[name]
        extra = f'  [runner-up: {runner_up[name].title()}]' if show_runner_up and name in runner_up else ''
        print(f' {name:<8} → {label} ({detail}){extra}')

print_archetypes(messages)
```

## Feature 8: The Final Report
Everything above, wrapped into one screenshot-ready report. Also works on edge inputs (empty file, one person).

```python
# AI-assisted: Claude helped with the box layout; each section function above is my own work.
def generate_report(lines):
    msgs, inf = parse_chat(lines)
    if len(msgs) == 0:
        print('=' * WIDTH)
        print(' GROUPDNA REPORT')
        print(' No messages found in this file. Nothing to analyse.')
        print('=' * WIDTH)
        return
    counts = count_per_person(msgs)
    days = (msgs[-1]['dt'].date() - msgs[0]['dt'].date()).days + 1
    first, last = msgs[0]['dt'], msgs[-1]['dt']
    names, grid = build_heatmap(msgs)
    owl_tags = {}
    people_list, assigned, runner_up = detect_archetypes(msgs)
    for name in people_list:
        if assigned[name][0] == 'THE NIGHT OWL':
            owl_tags[name] = '<- NIGHT OWL'
    top = max(counts.values())

    print('=' * WIDTH)
    print(f" GROUPDNA REPORT — \"{inf['group']}\"")
    print(f" {days} days • {len(msgs):,} messages • {len(counts)} member{'' if len(counts) == 1 else 's'}")
    print('=' * WIDTH)
    print(f" Period       : {first.strftime('%d %B %Y')} to {last.strftime('%d %B %Y')}")
    print_busiest(msgs)
    print(f" Extras       : {inf['system']} system, {sum(inf['media'].values())} media, {sum(inf['deleted'].values())} deleted messages")
    print('\n MESSAGES PER PERSON')
    for name in ranked(counts):
        c = counts[name]
        print(f' {name:<8} {make_bar(c, top):<20} {c:>5} ({c / len(msgs) * 100:4.1f}%)')
    print()
    print_heatmap(names, grid, owl_tags)
    print()
    print_top_words(msgs, 5)
    print()
    print_response_patterns(msgs)
    print()
    print_archetypes(msgs, show_runner_up=False)
    print('=' * WIDTH)
    print(' Generated by GroupDNA • Built with Python + NumPy')
    print('=' * WIDTH)

generate_report(lines)
```
